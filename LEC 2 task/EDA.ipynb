# EDA insights
we have 37 numeric features and 43 categorical features 
the categorical features needs to be splited into ordinal and nominal features

numeric features processing:
some unnecesary features were dropped (['Id','YearRemodAdd','MoSold','YrSold'])

Missing: we imputed the missing values (['LotFrontage']) with the mean of the column 

multicollinearity: we dropped 4 features (["TotalBsmtSF","GarageArea","1stFlrSF","GarageYrBlt"])

Zeros: The high number of zeros is normal for features like PoolArea or 2ndFlrSF, as most houses don't have them. This is often fine

Skewed: For skewed columns like MiscVal, should apply a log transformation (np.log1p) to normalize its distribution, which helps most models perform better. in feature engineering phase



catigorical features processing:

data needs imputation missing = none then zero

split to ordinal (['ExterQual', 'ExterCond', 'BsmtQual', 'BsmtCond', 'HeatingQC', 'KitchenQual', 'FireplaceQu', 'GarageQual', 'GarageCond', 'LandSlope', 'LotShape', 'PavedDrive'])
as the values are for quality evaluation and will be ordenaly encoded as int: 
 5      Ex	Excellent
 4      Gd	Good
 3      TA	Typical/Average
 2      Fa	Fair
 1      Po	Poor
 0      NA none

the rest are nominal will be hot encoded

like this we can avoide the curse of dimensionality
