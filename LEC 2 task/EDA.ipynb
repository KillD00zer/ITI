##### also in another way  I can say all features affect price so i will encode into sub-features ,to use ANOVA F-value correlation test then check for features that didnt do any correlation with price 

```python
from sklearn.feature_selection import f_regression
from sklearn.preprocessing import OneHotEncoder # or i will use pandas.get_dummies() easier.
```

```python
sub_features = pd.get_dummies(data[nominal_features], drop_first=True)
```

```python
X = sub_features
y = data['SalePrice']
f_statistic, p_values = f_regression(X, y)

sub_features_test = pd.DataFrame({
    'Feature': X.columns,
    'F-Statistic': f_statistic,
    'p-value': p_values
})

sub_features_test
```

###### make a list sub_features with P-value < 0.05 (the features that are significant) then check what features are not in this list 

```python
Fnames= list(sub_features_test["Feature"][sub_features_test["p-value"]<0.05])  # sub_features with P-value < 0.05

cleaned = [f.split("_")[0] for f in Fnames]  # Remove the suffix from the feature names to convert it into the real features names 

unique_features = sorted(set(cleaned))

print(unique_features)
print(nominal_features)
```

```python
diff = list(set(unique_features) ^ set(nominal_features))
print("Different elements:", diff)
```

##### now I will make the Maps needed for Ordinal encoding

```python
for i in ordinal_features:
    L= (list(data[i].unique()))
    print(str(i)+" : "+str(L))
```

```python
# maps of ordinal encoding

# 5-Level Quality/Condition Scale ('Poor' to 'Excellent')
quality_map_5_levels = {
    'Po': 1, 
    'Fa': 2, 
    'TA': 3, 
    'Gd': 4, 
    'Ex': 5, 
    'Nane':0,
    'nan': 0,
    np.nan: 0,
}


# Basement Finish Scale ('Unfinished' to 'Good Living Quarters')
bsmt_fin_map = {
    'Unf': 1, 
    'LwQ': 2, 
    'Rec': 3, 
    'BLQ': 4, 
    'ALQ': 5, 
    'GLQ': 6, 
    'Nane':0,
    'nan': 0,
    np.nan: 0,
}


# ## Unique Individual Maps ##

lot_shape_map = {'Reg': 3, 'IR1': 2, 'IR2': 1, 'IR3': 0}

land_slope_map = {'Gtl': 0, 'Mod': 1, 'Sev': 2}

bsmt_cond_map = {'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Nane': 0,  'nan': 0, np.nan: 0}

bsmt_exposure_map = {'No': 1, 'Mn': 2, 'Av': 3, 'Gd': 4, 'Nane': 0, 'nan': 0, np.nan: 0}

garage_finish_map = {'Unf': 1, 'RFn': 2, 'Fin': 3, 'Nane': 0, 'nan': 0, np.nan: 0}

paved_drive_map = {'N': 0, 'P': 1, 'Y': 2}
```

##### lets try the encoding 

```python
encodata = pd.DataFrame()

# Apply the common 5-Level Quality Map
encodata['ExterQual'] = data['ExterQual'].replace(quality_map_5_levels)
encodata['ExterCond'] = data['ExterCond'].replace(quality_map_5_levels)
encodata['BsmtQual'] = data['BsmtQual'].replace(quality_map_5_levels)
encodata['BsmtCond'] = data['BsmtCond'].replace(quality_map_5_levels)
encodata['HeatingQC'] = data['HeatingQC'].replace(quality_map_5_levels)
encodata['KitchenQual'] = data['KitchenQual'].replace(quality_map_5_levels)
encodata['FireplaceQu'] = data['FireplaceQu'].replace(quality_map_5_levels)
encodata['GarageQual'] = data['GarageQual'].replace(quality_map_5_levels)
encodata['GarageCond'] = data['GarageCond'].replace(quality_map_5_levels)
encodata['PoolQC'] = data['PoolQC'].replace(quality_map_5_levels)

# Apply the Basement Finish Map
encodata['BsmtFinType1'] = data['BsmtFinType1'].replace(bsmt_fin_map)
encodata['BsmtFinType2'] = data['BsmtFinType2'].replace(bsmt_fin_map)

# Apply the Unique/Individual Maps
encodata['LotShape'] = data['LotShape'].replace(lot_shape_map)
encodata['LandSlope'] = data['LandSlope'].replace(land_slope_map)
encodata['BsmtExposure'] = data['BsmtExposure'].replace(bsmt_exposure_map)
encodata['GarageFinish'] = data['GarageFinish'].replace(garage_finish_map)
encodata['PavedDrive'] = data['PavedDrive'].replace(paved_drive_map)
```
